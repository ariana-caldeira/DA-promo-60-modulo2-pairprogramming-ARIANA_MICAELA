```python
# antes de empezar importamos las librerías que vamos a usar. 
# Importar librerías para web scraping y manipulación de datos
# -----------------------------------------------------------------------
from bs4 import BeautifulSoup
import requests

# Importar librerías para manipulación y análisis de datos
# -----------------------------------------------------------------------
import pandas as pd

# Importar librerías para procesamiento de texto
# -----------------------------------------------------------------------
import re
```

```python
url = 'https://www.esquire.com/es/actualidad/libros/g63833331/libros-mas-vendidos-2025-espana/'
```

```python
rq_esquire = requests.get(url)

print("La respuesta de la petición es:",rq_esquire.status_code)
```

```python
sopa = BeautifulSoup(rq_esquire.content, 'html.parser')
print(sopa.prettify())
```

```python
sopa.find_all('h2')
```

```python
sopa.find_all('h2', class_="css-1msffwn e1bddzxd0")[0].text
```

```python
lista_esquire_2025 = []

for libro in sopa.find_all('h2', class_="css-1msffwn e1bddzxd0"):
    lista_esquire_2025.append(libro.text.split(' ', 1))
```

```python
lista_esquire_2025
```

```python
editoras = []
libro = []

for e in lista_esquire_2025:
    editoras.append(e [0])
    libro.append(e [1])
```

```python
editoras
```

```python
libro
```

```python
libros_split = []

for l in libro:
    libros_split.append(l.split(' - '))
```

```python
libros_split
```

```python
titulo_libro = []
autor = []


for e in libros_split:
    titulo_libro.append(e [0])
    autor.append(e [1])
```
